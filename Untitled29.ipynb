```python
import numpy as np
import evaluate
from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

# 1. LOAD REAL MOVIE REVIEW DATASET (IMDb)
print("--- Loading IMDb Dataset ---")
# Yeh automatic internet se real movie reviews load karega
raw_datasets = load_dataset("imdb")

# Dataset bohot bada hai (25,000 reviews), isliye sirf ek chota part use karenge
# taaki aapke system par jaldi run ho jaye.
train_dataset = raw_datasets["train"].shuffle(seed=42).select(range(1000))
test_dataset = raw_datasets["test"].shuffle(seed=42).select(range(200))

# 2. TOKENIZATION
MODEL_NAME = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def preprocess_function(examples):
    # Movie reviews lambe hote hain, isliye max_length=256 rakha hai
    return tokenizer(examples["text"], truncation=True, padding="max_length", max_length=256)

print("--- Tokenizing Dataset ---")
tokenized_train = train_dataset.map(preprocess_function, batched=True)
tokenized_test = test_dataset.map(preprocess_function, batched=True)

# 3. COMPUTE METRICS FUNCTION
def compute_metrics(eval_pred):
    metric_acc = evaluate.load("accuracy")
    metric_prec = evaluate.load("precision")
    metric_rec = evaluate.load("recall")
    metric_f1 = evaluate.load("f1")

    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    return {
        "accuracy": metric_acc.compute(predictions=predictions, references=labels)["accuracy"],
        "precision": metric_prec.compute(predictions=predictions, references=labels)["precision"],
        "recall": metric_rec.compute(predictions=predictions, references=labels)["recall"],
        "f1": metric_f1.compute(predictions=predictions, references=labels)["f1"],
    }

# 4. LOAD TRANSFORMER MODEL (For Binary Classification: Positive/Negative)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2)

# 5. DEFINE TRAINING ARGUMENTS
training_args = TrainingArguments(
    output_dir="./results_movie",
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=2,              # 2 Epochs kafi hain demo ke liye
    weight_decay=0.01,
    evaluation_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    logging_steps=20
)

# 6. INITIALIZE TRAINER AND START FINE-TUNING
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_test,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics,
)

print("--- Starting Model Training on Movie Reviews ---")
trainer.train()

# 7. EVALUATION REPORT
print("\n--- Running Final Evaluation ---")
eval_results = trainer.evaluate()

print("\n===== FINAL MOVIE REVIEW METRICS REPORT =====")
print(f"Accuracy  : {eval_results['eval_accuracy']:.4f}")
print(f"Precision : {eval_results['eval_precision']:.4f}")
print(f"Recall    : {eval_results['eval_recall']:.4f}")
print(f"F1-Score  : {eval_results['eval_f1']:.4f}")
print("==============================================")
```
